## 4. Synthesis & reflection

*Bring the probe and the observation together into one account. What does this mean for trustworthy, reproducible research — and what will you change about how you use and verify AI in your own work?*

The reproducibility probe showed that the models were consistent on a simple, checkable statistics task, but the observation showed that people can still become trusting very quickly once a few parts of an answer seem correct. My participant checked the mean and missing percentage, then accepted the median and mode without verifying them separately. That made me realize that reproducibility does not automatically mean trustworthiness, especially for more complicated tasks where there may not be an obvious ground truth.

For my own work, I want to be more deliberate about separating using an LLM for help from treating its output as evidence. I will still use AI to explain concepts, troubleshoot code, and help me get started, but I want to verify important claims with the original data, calculations, or sources instead of assuming that a confident or consistent answer is correct.
